# Using Mistral.AI (optional exercise)
If you have access to [Mistral.AI](https://mistral.ai/), you can also use Bia-Bob together with Mistral models. 
Just [acquire an API Key](https://admin.mistral.ai/organization/api-keys) and use a text-editor to edit the  `.env` file in the current directory. Enter your API key like this:
```
MISTRAL_API_KEY="<your key>"
```

You can then initialize bob like this:

````python
# load secret API key. You must unpack the contents of api_key.zip 
# into the same folder before going ahead.
from dotenv import load_dotenv
load_dotenv(override=True)

import os
from bia_bob import bob
bob.initialize(endpoint='https://api.mistral.ai/v1', 
               model='codestral-2508', 
               vision_model='codestral-2508', 
               api_key=os.environ.get('MISTRAL_API_KEY'))
````

Your prompts wwill then be sent to the Mistral server in France.

````python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 100)
y = sigmoid(x)

plt.plot(x, y)
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel('sigmoid(x)')
plt.grid(True)
plt.show()
````
